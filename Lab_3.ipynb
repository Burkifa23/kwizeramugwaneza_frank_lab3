### Part 1.2 — Hidden layers and activations: why depth helps

Now build a tiny network **by hand**: 1 input -> a hidden layer of 8 neurons with `tanh` ->
1 output. Still NumPy, still manual gradients.


```python
# Reshape
X = x.reshape(-1, 1)
Y = y.reshape(-1, 1)


# Function
def train_network(X, Y, lr=0.05, steps=3000, use_activation=True):
    np.random.seed(42)

    # Initialize parameters
    W1 = np.random.randn(1, 8) * 0.5
    b1 = np.zeros(8)
    W2 = np.random.randn(8, 1) * 0.5
    b2 = np.zeros(1)

    losses = []

    for i in range(steps):
        # Forward pass
        z1 = X @ W1 + b1
        h = np.tanh(z1) if use_activation else z1
        y_hat = h @ W2 + b2
        loss = np.mean((y_hat - Y) ** 2)

        # Backward pass
        d_yhat = 2 * (y_hat - Y) / len(Y)
        dW2 = h.T @ d_yhat
        db2 = d_yhat.sum(axis=0)
        dh = d_yhat @ W2.T

        # Toggle derivative based on activation
        dz1 = dh * (1 - np.tanh(z1) ** 2) if use_activation else dh

        dW1 = X.T @ dz1
        db1 = dz1.sum(axis=0)

        # Update parameters
        W1 -= lr * dW1
        b1 -= lr * db1
        W2 -= lr * dW2
        b2 -= lr * db2

        losses.append(loss)

    return losses, y_hat


losses_tanh, y_hat_tanh = train_network(X, Y, use_activation=True)
# Train WITHOUT Tanh
losses_no_tanh, y_hat_no_tanh = train_network(X, Y, use_activation=False)


# Plot
plt.plot(losses_tanh)
plt.title("Loss Curve (With TANH)")
plt.xlabel("Step")
plt.ylabel("MSE")
plt.show()

plt.scatter(X.flatten(), Y.flatten())
plt.plot(X.flatten(), y_hat_tanh.flatten())
plt.title("Final Fit (With TANH)")
plt.show()

# PLOT WITHOUT Tanh
plt.plot(losses_no_tanh)
plt.title("Loss Curve (Without TANH)")
plt.xlabel("Step")
plt.ylabel("MSE")
plt.show()

plt.scatter(X.flatten(), Y.flatten())
plt.plot(X.flatten(), y_hat_no_tanh.flatten())
plt.title("Final Fit (Without TANH)")
plt.show()
```

### Part 1.3 — The learning rate: too cold, too hot, just right

The single most important hyperparameter. See its three regimes with your own eyes.


```python
# Reshaped
X = x.reshape(-1, 1)
Y = y.reshape(-1, 1)


def train_toy(lr):
    np.random.seed(42)

    W1 = np.random.randn(1, 8) * 0.5
    b1 = np.zeros(8)
    W2 = np.random.randn(8, 1) * 0.5
    b2 = np.zeros(1)

    losses = []

    for i in range(3000):
        # Forward pass (with tanh)
        z1 = X @ W1 + b1
        h = np.tanh(z1)
        y_hat = h @ W2 + b2

        loss = np.mean((y_hat - Y) ** 2)
        losses.append(loss)

        # Backward pass
        d_yhat = 2 * (y_hat - Y) / len(Y)
        dW2 = h.T @ d_yhat
        db2 = d_yhat.sum(axis=0)
        dh = d_yhat @ W2.T
        dz1 = dh * (1 - np.tanh(z1) ** 2)
        dW1 = X.T @ dz1
        db1 = dz1.sum(axis=0)

        # Parameter update
        W1 -= lr * dW1
        b1 -= lr * db1
        W2 -= lr * dW2
        b2 -= lr * db2

    return losses


# Train lr=0.001
losses_slow = train_toy(lr=0.001)
# Train lr=0.05
losses_good = train_toy(lr=0.05)
# Train lr=1.0
losses_fast = train_toy(lr=1.0)

# As 1 Plot
plt.figure(figsize=(10, 6))
plt.plot(losses_slow, label="lr = 0.001 (Too Cold)", color="blue")
plt.plot(losses_good, label="lr = 0.05 (Just Right)", color="green")
plt.plot(losses_fast, label="lr = 1.0 (Too Hot)", color="red", alpha=0.7)

plt.title("The Effect of Learning Rate on Training Loss")
plt.xlabel("Step")
plt.ylabel("MSE Loss (Log Scale)")
plt.yscale("log")
plt.ylim(1e-2, 10)
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```

**Student Reasoning — Learning rate**
Describe what each of the three curves shows. Why does a too-large learning rate make the loss explode or oscillate instead of just "learning faster"?

> The Learning Rate of 0.001 is too cold because the loss decreases very slowly and it it is not as smooth.
> The Learning Rate of 0.05 is just right because the loss drops rapidly yet smoothly.
> The Learning Rate of 1.0 is too hot because the it seems like it is shooting up to the top.

> A large learning rates introduce a kind of instability or oscillation near the minimum. This causes the optimizer to settle in a region that minimizes oscillations, which might not be the best minimum but is a stable point given the high learning rate which in turn leads to a suboptimal convergence point compared to using a lower or “good” learning rate.
